Проверка и очистка пустых ячеек

```python
import pandas as pd
from pathlib import Path

os.makedirs('data/processed', exist_ok=True)
PROJECT_ROOT = Path(os.getcwd())
RAW_DIR = PROJECT_ROOT / 'data' / 'raw'
PROCESSED_DIR = PROJECT_ROOT / 'data' / 'processed'

df_raw_district_prices_monthly = pd.read_csv(RAW_DIR / 'district_prices_monthly.csv')

# Проверка данных
print(f"Проверка данных")
print(f"Строк: {len(df_raw_district_prices_monthly)}")
print(f"Столбцов: {len(df_raw_district_prices_monthly.columns)}")
print(f"Пустых значений: {df_raw_district_prices_monthly.isnull().sum().sum()}")

# Приведение к формату
df_clean_district_prices_monthly = df_raw_district_prices_monthly.copy()
df_clean_district_prices_monthly['year_month'] = pd.to_datetime(df_clean_district_prices_monthly['year_month'])
```

Проверим пустые ячейки

```python
print("Пропуски по колонкам:")
print(df_clean_district_prices_monthly.isnull().sum()[df_clean_district_prices_monthly.isnull().sum() > 0])

print("\nВ каком месяце эти пропуски:")
print(df_clean_district_prices_monthly[df_clean_district_prices_monthly['secondary_mom_change_pct'].isnull()]['year_month'].value_counts())
```

Все пропуски относятся к месяцу 2020-01 - первому месяцу в датасете, для которого нет предыдущего месяца, чтобы посчитать % изменения.

```python
df_clean_district_prices_monthly.to_csv('data/processed/clean_district_prices_monthly.csv', index=False)
```
